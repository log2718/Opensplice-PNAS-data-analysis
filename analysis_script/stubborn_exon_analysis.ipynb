```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()   # if running from repo root in notebook
# if this fails, use your explicit repo path instead

INPUT_CSV = ROOT / '..' / "outputs" / "stubborn_75_seq_struct_components.csv"
OUT_DIR = ROOT / '..' /"outputs" / "plots" / "stubborn_75_seq_struct"
OUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(INPUT_CSV)

print("Number of exons:", df["exon_id"].nunique())
print("Number of rows:", len(df))
```

```python
for exon_id, g in df.groupby("exon_id"):
    g = g.dropna(subset=["seq_sumdiff", "struct_sumdiff"])

    if len(g) == 0:
        continue

    plt.figure(figsize=(7, 6))

    plt.scatter(
        g["seq_sumdiff"],
        g["struct_sumdiff"],
        alpha=0.7,
        s=40,
    )

    # reference lines
    plt.axhline(0, linestyle="--", linewidth=1)
    plt.axvline(0, linestyle="--", linewidth=1)

    plt.xlabel("Sequence balance (inclusion minus skipping)")
    plt.ylabel("Structure balance (inclusion minus skipping)")
    plt.title(f"{exon_id}: sequence vs structure contribution")

    plt.tight_layout()

    out_path = OUT_DIR / f"{exon_id}_seq_struct_filters_graph.png"
    plt.savefig(out_path, dpi=200, bbox_inches="tight")
    plt.close()

print(f"Saved plots to: {OUT_DIR}")
```

```python
for exon_id, g in df.groupby("exon_id"):
    g = g.dropna(subset=["seq_sumdiff", "struct_sumdiff", "logit_psi"])

    if len(g) == 0:
        continue

    plt.figure(figsize=(7, 6))

    sc = plt.scatter(
        g["seq_sumdiff"],
        g["struct_sumdiff"],
        c=g["logit_psi"],
        alpha=0.75,
        s=40,
    )

    plt.axhline(0, linestyle="--", linewidth=1)
    plt.axvline(0, linestyle="--", linewidth=1)

    plt.xlabel("Sequence balance (inclusion minus skipping)")
    plt.ylabel("Structure balance (inclusion minus skipping)")
    plt.title(f"{exon_id}: sequence vs structure contribution")

    cbar = plt.colorbar(sc)
    cbar.set_label("Measured logit PSI")

    plt.tight_layout()

    out_path = OUT_DIR / f"{exon_id}_seq_struct_filters_graph.png"
    plt.savefig(out_path, dpi=200, bbox_inches="tight")
    plt.close()

print(f"Saved plots to: {OUT_DIR}")
```
