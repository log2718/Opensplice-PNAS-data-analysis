```python
#copying of manually inspected nonlinear exons

nonlinear_exons = [
    "ACACA_e2",
    "ADAM32_e10",
    "AGBL2_e17",
    "AKAP3_e3",
    "ASZ1_e5",
    "BORA_e3",
    "COL4A3_e18",
    "COL9A3_e3",
    "COL11A2_e62",
    "EYA4_e19",
    "MEIKIN_e11",
    "MS4A3_e3",
    "NBN_e3",
    "SNRNP70_e8",
]
```

```python
from pathlib import Path
import shutil

src_dir = Path("../outputs/plots")
dst_dir = src_dir / "nonlinear"

dst_dir.mkdir(exist_ok=True)

copied = 0

for exon_id in nonlinear_exons:
    matches = list(
        src_dir.glob(f"{exon_id}_with_ss_pretuner_vs_measured*.png")
    )

    if not matches:
        print("Missing:", exon_id)
        continue

    for file in matches:
        shutil.copy2(file, dst_dir / file.name)
        copied += 1

print("Copied:", copied)
```
